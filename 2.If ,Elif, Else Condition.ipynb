```python
# IF & Elif Conditions
# To check Marks and Grade A = 80, B = 60 till 80, C = 40 till 60 D less than 40.
marks=int(input("Enter your Marks:"))
if(marks<40):
    print("Need Improvement")
elif(marks<60):    
    print("Average")
elif(marks<80):   
    print("Good")
else:
    print("Excellent Keep it up")






```
