```python
# 10 Create a function that tells elegibility of marriage for male and female according to thei
# r age limit like 21 for male and 18 for female

#class ElegiblityForMarriage:
def Elegible():
        lv_gen = input("Your Gender:")
        lv_age = int(input("Your Age:"))
        if ((lv_gen=='Male') and (lv_age>21)):
            print("ELIGIBLE")
        elif ((lv_gen=='Female') and (lv_age>18)):  
             print("ELIGIBLE")
        else:
            print("NOT ELIGIBLE")
Elegible()           
```

```python
# 13 calculate the percentage of your 10th mark
#class FindPercent():
def percentage():
        sub1=int(input("Subject1="))
        sub2=int(input("Subject2="))
        sub3=int(input("Subject3="))
        sub4=int(input("Subject4="))
        sub5=int(input("Subject5="))
        total = sub1+sub2+sub3+sub4+sub5
        percent = (total / 500) * 100
        print("Subject1= ",sub1)
        print("Subject2= ",sub2)
        print("Subject3= ",sub3)
        print("Subject4= ",sub4)
        print("Subject5= ",sub5)
        print("Total : ",total)
        print("Percentage :",percent)
percentage()
```
