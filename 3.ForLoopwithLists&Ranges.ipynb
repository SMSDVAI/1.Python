```python
# For Loop Condition Statement
# Multiple Inputs to check 1 set of condition
# Syntax for temp in mainvariable:
# Lists   print("Weight Check")

lists=[10,20,30,40,50,60,80,90,100]
print("Weight Check")
for weight in lists:   
    if(weight<40):
     print("Your Weight:",weight)
     print("Under Weight")
    elif(weight<60):
     print("Your Weight:",weight)
     print("Corrrect Weight Your Selected")
    elif(weight<80):
     print("Your Weight:",weight)
     print("You need to Reduce 10 KG")
    else:
     print("Your Weight:",weight)
     print("Your Overweight & Unfit for this Selection") 



```
